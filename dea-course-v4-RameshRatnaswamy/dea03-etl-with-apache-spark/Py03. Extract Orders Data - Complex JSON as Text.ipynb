## Extract Data From the Orders JSON File
1. Query Orders File using JSON Format
1. Query Orders File using TEXT Format
1. Create Orders Table in Bronze Schema

### 1. Query Orders File using JSON Format

```python
df = spark.read.json("/Volumes/gizmobox/landing/operational_data/orders")
display(df)
```

### 2. Query Orders File using TEXT Format

```python
df = spark.read.text("/Volumes/gizmobox/landing/operational_data/orders")
display(df)
```

### 3. Create Orders View in Bronze Schema

```python
df.writeTo('gizmobox.bronze.py_orders').createOrReplace()
```

```python
%sql
SELECT * FROM gizmobox.bronze.py_orders;
```
