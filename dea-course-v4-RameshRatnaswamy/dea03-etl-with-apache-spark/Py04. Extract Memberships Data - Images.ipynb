## Extract Data From the Memberships - Image Files
1. Query Memberships File using binaryFile Format
1. Create Memberships Table in Bronze Schema

### 1. Query Memberships Folder using binaryFile Format

```python
%fs ls '/Volumes/gizmobox/landing/operational_data/memberships'
```

```python
df = spark.read.format('binaryFile').load('/Volumes/gizmobox/landing/operational_data/memberships/2024-10/*.png')
display(df)
```

```python
df = spark.read.format('binaryFile').load('/Volumes/gizmobox/landing/operational_data/memberships/*/*.png')
display(df)
```

### 2. Create Memberships Table in Bronze Schema

```python
df.writeTo('gizmobox.bronze.py_memberships').createOrReplace()
```

```python
%sql
SELECT * FROM gizmobox.bronze.py_memberships;
```
